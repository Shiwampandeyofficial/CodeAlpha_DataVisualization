```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

url = "https://raw.githubusercontent.com/plotly/datasets/master/gapminder_with_codes.csv"
df = pd.read_csv(url)
df.head()
```

```python
d07 = df[df['year'] == 2007]
yearly = df.groupby('year')['lifeExp'].mean().reset_index()

plt.figure(figsize=(8,5))
sns.lineplot(x='year', y='lifeExp', data=yearly, marker='o', color='teal')
plt.title("Average Life Expectancy of the World (1952-2007)")
plt.xlabel("Year")
plt.ylabel("Life Expectancy (years)")
plt.show()
```

***Insights:***

> Global average life expectancy rose steadily from about 49 years in 1952 to about 67 years in 2007, a gain of roughly 18 years in 55 years.



```python
d07 = df[df['year'] == 2007]

plt.figure(figsize=(9,6))
sns.scatterplot(x='gdpPercap', y='lifeExp', size='pop', hue='continent',
                data=d07, sizes=(30,1200), alpha=0.7)
plt.xscale('log')
plt.title("Income vs Life Expectancy (2007)")
plt.legend(bbox_to_anchor=(1.02,1), loc='upper left')
plt.tight_layout()
plt.show()
```

***Insights:***

>  Countries with higher income per person tend to have longer life expectancy. The rise is steep at low incomes and flattens at high incomes, which means extra money helps most in poorer countries. Asia and Africa contain the large bubbles and the widest range.



```python
cont = d07.groupby('continent')['lifeExp'].mean().sort_values(ascending=False).reset_index()

plt.figure(figsize=(8,5))
sns.barplot(x='continent', y='lifeExp', data=cont, hue='continent', palette='viridis', legend=False)
plt.title("Average Life Expectancy by Continent (2007)")
plt.show()
```

**Insights**


> In 2007, Oceania (about 81 years) and Europe (about 78 years) had the highest average life expectancy, while Africa (about 55 years) had the lowest.



```python
plt.figure(figsize=(8,5))
sns.boxplot(x='continent', y='lifeExp', data=d07, hue='continent', palette='Set2', legend=False)
plt.title("Life Expectancy Spread by Continent (2007)")
plt.show()
```

**Insights**

> Norway, Kuwait and Singapore had the highest income per person in 2007, each above 47,000 dollars. The top 10 mixes oil-rich countries and developed economies.



```python
top10 = d07.sort_values('gdpPercap', ascending=False).head(10)

plt.figure(figsize=(9,5))
sns.barplot(x='gdpPercap', y='country', data=top10, hue='country', palette='mako', legend=False)
plt.title("Top 10 Richest Countries by GDP per Capita (2007)")
plt.tight_layout()
plt.show()
```

**Insights**


> Norway, Kuwait and Singapore had the highest income per person in 2007, each above 47,000 dollars. The top 10 mixes oil-rich countries and developed economies.


```python
pc = df.groupby(['year','continent'])['pop'].sum().reset_index()
pc['pop'] = pc['pop'] / 1e9

plt.figure(figsize=(8,5))
sns.lineplot(x='year', y='pop', hue='continent', data=pc, marker='o')
plt.title("Population by Continent (1952-2007)")
plt.ylabel("Population (billions)")
plt.show()
```

**Insights**


> Asia's population grew the most, from about 1.4 billion in 1952 to about 3.8 billion in 2007. Africa grew fastest in percentage terms (about 0.24 to 0.93 billion), while Europe grew slowly.


```python
plt.figure(figsize=(6,5))
sns.heatmap(d07[['lifeExp','pop','gdpPercap']].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation (2007)")
plt.show()
```

**Insights**

> GDP per capita and life expectancy are strongly positively correlated (0.68). Population has almost no relationship with either one (0.05 and -0.06).



## Conclusion
---

- The world became healthier: average life expectancy rose by about 18 years from 1952 to 2007.
- Higher income is strongly linked to longer life, but the benefit levels off at high incomes.
- Africa still has the lowest life expectancy and large differences between countries.
- Asia saw the biggest population growth.
